```python
import pandas as pd
```

```python
arc_df: pd.DataFrame = None

for exp in loaded_dict:
    exp_arc = loaded_dict.get(exp).get('geometry_data_stl_suitable_arc')
    exp_df = pd.DataFrame(exp_arc)

    exp_df['Experiment_ID'] = exp.split('_')[1]
    exp_df['Tube_Section'] = exp_df['Tube_Section'].str.split('_').str[1]

    if arc_df is None:
        arc_df = exp_df
    else:
        arc_df = pd.concat([arc_df, exp_df], ignore_index=True)
```

```python
arc_df = arc_df[['Experiment_ID', 'Tube_Section' , 'X [mm]', 'Y [mm]', 'Z [mm]']].reset_index(drop=True)
arc_df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

axis = 'Z [mm]'
arc_df_exp_1 = arc_df[arc_df['Experiment_ID'] == '1'][['Tube_Section', axis]]

df = arc_df_exp_1.copy()
df['Tube_Section'] = df['Tube_Section'].astype(int)

plt.figure(figsize=(12, 6))

for section in sorted(df['Tube_Section'].unique()):
    section_data = df[df['Tube_Section'] == section][axis].values
    n = len(section_data)
    # Create sinusoidal jitter for x-axis displacement
    jitter = np.sin(np.linspace(0, np.pi, n)) * 0.3  # amplitude of 0.3 for clarity
    plt.scatter(section_data + jitter, 
                np.full_like(section_data, section), 
                s=10, alpha=0.7, c=section_data, cmap='viridis')

plt.xlabel('X [mm] + sinusoidal jitter')
plt.ylabel('Tube_Section')
plt.title('Sinusoidal Scatter Plot per Tube Section')
plt.colorbar(label='X [mm]')
plt.show()

# Copy and ensure Tube_Section is numeric
df = arc_df_exp_1.copy()
df['Tube_Section'] = df['Tube_Section'].astype(int)
df = df.sort_values('Tube_Section')

# --- 1. Scatter plot ---
plt.figure(figsize=(12, 5))
plt.scatter(df[axis], df['Tube_Section'], c=df[axis], cmap='viridis', s=10)
plt.xlabel('X [mm]')
plt.ylabel('Tube_Section')
plt.title('Scatter Plot: X vs Tube_Section')
plt.colorbar(label='X [mm]')
plt.show()

# --- 2. Strip plot (jittered) ---
plt.figure(figsize=(12, 5))
sns.stripplot(x='Tube_Section', y=axis, data=df, jitter=0.3, size=3)
plt.xlabel('Tube_Section')
plt.ylabel('X [mm]')
plt.title('Strip Plot: X distribution per Tube_Section')
plt.show()

# --- 3. Box plot ---
plt.figure(figsize=(12, 5))
sns.boxplot(x='Tube_Section', y=axis, data=df)
plt.xlabel('Tube_Section')
plt.ylabel('X [mm]')
plt.title('Box Plot: X distribution per Tube_Section')
plt.show()

# --- 4. Line plot of mean X per Tube_Section ---
mean_x = df.groupby('Tube_Section')[axis].min()
plt.figure(figsize=(12, 5))
plt.plot(mean_x.index, mean_x.values, marker='o')
plt.xlabel('Tube_Section')
plt.ylabel('Mean X [mm]')
plt.title('Line Plot: Mean X per Tube_Section')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

axis = 'X [mm]'
arc_df_exp_1 = arc_df[arc_df['Experiment_ID'] == '1'][['Tube_Section', axis]]

df = arc_df_exp_1.copy()
df['Tube_Section'] = df['Tube_Section'].astype(int)

plt.figure(figsize=(12, 6))

for section in sorted(df['Tube_Section'].unique()):
    section_data = df[df['Tube_Section'] == section][axis].values
    n = len(section_data)
    
    # Sinusoidal jitter for display (amplitude can be adjusted)
    jitter = np.sin(np.linspace(0, np.pi, n)) * 0.3
    x_vals = section_data + jitter
    y_vals = np.full_like(x_vals, section)
    
    # Plot points
    plt.scatter(x_vals, y_vals, s=10, alpha=0.7, c=section_data, cmap='viridis')

plt.xlabel('X [mm] + sinusoidal jitter')
plt.ylabel('Tube_Section')
plt.title('Sinusoidal Scatter Plot per Tube Section')
plt.colorbar(label='X [mm]')
plt.show()
```
